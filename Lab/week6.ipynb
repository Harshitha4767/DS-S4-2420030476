```python
#1. Measures of Central Tendency
import pandas as pd
from sklearn.datasets import load_iris

# Load Iris dataset
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)

print("First 5 rows of dataset:")
print(df.head())

# Mean
mean_values = df.mean()
print("\nMean values:")
print(mean_values)

# Median
median_values = df.median()
print("\nMedian values:")
print(median_values)

# Mode
mode_values = df.mode().iloc[0]
print("\nMode values:")
print(mode_values)
```

```python
#2. Measures of Dispersion
# Range
range_values = df.max() - df.min()
print("\nRange values:")
print(range_values)

# Variance
variance_values = df.var()
print("\nVariance values:")
print(variance_values)

# Standard Deviation
std_values = df.std()
print("\nStandard Deviation values:")
print(std_values)

# Interquartile Range (IQR)
Q1 = df.quantile(0.25)
Q3 = df.quantile(0.75)
IQR = Q3 - Q1
print("\nInterquartile Range (IQR):")
print(IQR)
```

```python
#3. Shape of Distribution
# Skewness
skewness_values = df.skew()
print("\nSkewness values:")
print(skewness_values)

# Kurtosis
kurtosis_values = df.kurt()
print("\nKurtosis values:")
print(kurtosis_values)
```

```python
#4. Correlation & Association
# Pearson Correlation
pearson_corr = df.corr(method='pearson')
print("\nPearson Correlation:")
print(pearson_corr)

# Spearman Correlation
spearman_corr = df.corr(method='spearman')
print("\nSpearman Correlation:")
print(spearman_corr)

# Chi-Square Test (example with categorical target)
from scipy.stats import chi2_contingency
import numpy as np

# Convert target into categorical DataFrame
target = pd.Series(iris.target, name="species")
contingency_table = pd.crosstab(target, pd.cut(df['sepal length (cm)'], bins=3))
chi2, p, dof, expected = chi2_contingency(contingency_table)

print("\nChi-Square Test:")
print(f"Chi2: {chi2}, p-value: {p}")
```

```python
#5. Hypothesis Testing
from scipy.stats import ttest_ind, f_oneway, norm

# Example: T-test between two species
setosa = df.iloc[iris.target == 0, 0]  # sepal length of setosa
versicolor = df.iloc[iris.target == 1, 0]  # sepal length of versicolor
t_stat, p_val = ttest_ind(setosa, versicolor)
print("\nT-test between Setosa and Versicolor sepal length:")
print(f"T-statistic: {t_stat}, p-value: {p_val}")

# ANOVA across all species
group1 = df.iloc[iris.target == 0, 0]
group2 = df.iloc[iris.target == 1, 0]
group3 = df.iloc[iris.target == 2, 0]
anova_stat, anova_p = f_oneway(group1, group2, group3)
print("\nANOVA across species (sepal length):")
print(f"F-statistic: {anova_stat}, p-value: {anova_p}")

# Z-test (sample mean vs population mean)
sample_mean = setosa.mean()
pop_mean = df['sepal length (cm)'].mean()
pop_std = df['sepal length (cm)'].std()
z_score = (sample_mean - pop_mean) / (pop_std / (len(setosa)**0.5))
print("\nZ-test for Setosa sepal length vs population mean:")
print(f"Z-score: {z_score}")
```

```python
#6. Outlier Detection
import numpy as np
from scipy import stats

def grubbs_test(series, alpha=0.05):
    data = series.values
    n = len(data)
    mean = np.mean(data)
    std_dev = np.std(data, ddof=1)
    
    # Find the most extreme value
    abs_diff = abs(data - mean)
    max_diff_index = np.argmax(abs_diff)
    G_calculated = abs_diff[max_diff_index] / std_dev
    
    # Critical value
    t_dist = stats.t.ppf(1 - alpha/(2*n), n-2)
    numerator = (n-1) * t_dist
    denominator = np.sqrt(n) * np.sqrt(n-2 + t_dist**2)
    G_critical = numerator / denominator
    
    return {
        "G_calculated": G_calculated,
        "G_critical": G_critical,
        "outlier_value": data[max_diff_index] if G_calculated > G_critical else None
    }

# Example usage with Iris sepal length
result = grubbs_test(df['sepal length (cm)'])
print("\nGrubbs’ Test result:")
print(result)

```
