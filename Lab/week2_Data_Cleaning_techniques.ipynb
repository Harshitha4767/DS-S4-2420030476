```python
#A) Handling missing values
#1. Mean/Meadian/Mode Imputation:

import pandas as pd
import numpy as np

#Sample Dataset
df = pd.DataFrame({ 'Age':[25, 30, np.nan, 40, 35],
    'Department': ['HR', 'Finance', 'Finance', np.nan, 'IT']})

#Display original dataset
print("Original Dataset (With Missing Value):")
print(df)

#Mean for numeric
df['Age'] = df['Age'].fillna(df['Age'].mean())

#Mode for categorical
df['Department'] = df['Department'].fillna(df['Department'].mode()[0])
print(df)
```

```python
#2. Forward and Backward fill

import pandas as pd
import numpy as np

#Sample Dataset
df = pd.DataFrame({ 'Age':[25, 30, np.nan, 40, 35],
    'Department': ['HR', 'Finance', 'Finance', np.nan, 'IT']})

#Display original dataset
print("Original Dataset (With Missing Value):")
print(df)

#Forward fill(use previous value)
df_ffill = df.copy()
df_ffill.ffill(inplace = True)
print(df_ffill)

#Backward fill(use next value)
df_bfill = df.copy()
df_bfill.bfill(inplace = True)
print(df_bfill)
```

```python
#3.a) deletion(drop rows and columns)

import pandas as pd
import numpy as np

#Sample Dataset(in department null values should'nt be given)
df = pd.DataFrame({ 'Age':[25, 30, np.nan, 40, 35],
    'Department': ['HR', 'Finance', 'Finance', 'IT', 'IT']})

#Display original dataset
print("Original Dataset (With Missing Value):")
print(df)

#Drop rows with missing values
df_drop_rows = df.dropna()
print("After dropping rows:\n", df_drop_rows)

#Drop columns with missing values
df_drop_cols = df.dropna(axis=1)
print("/nAfter dropping columns:\n", df_drop_cols)
```

```python
#3.b.a Removing duplicates(Exact match removal)

import pandas as pd

#Sample dataset with duplicates
df = pd.DataFrame({
    'ID': [1, 2, 2, 3, 4, 4],
    'Name': ['Alice', 'Bob', 'Bob', 'Charlie', 'David', 'David'],
    'Age': [25, 30, 30, 35, 40, 40]
})

print("Original data:\n", df)

#Remove duplicates
df_exact = df.drop_duplicates()
print("\nAfter exact match removal:\n", df_exact)
```

```python
#3.b.b Removing duplicates(Subset_based removal)

import pandas as pd

#Sample dataset with duplicates
df = pd.DataFrame({
    'ID': [1, 2, 2, 3, 4, 4, 5],
    'Name': ['Alice', 'Bob', 'Bob', 'Charlie', 'David', 'David', 'David'],
    'Age': [25, 30, 30, 35, 40, 40, 40]
})

print("Original data:\n", df)

#Remove duplicateds based only on 'ID'
df_subset_id = df.drop_duplicates(subset=['ID'])
print("\nAfter subset based removal (ID) :\n", df_subset_id)

#Remove duplicateds based only on 'Name'
df_subset_name = df.drop_duplicates(subset=['Name'])
print("\nAfter subset based removal (Name):\n", df_subset_name)
```

```python
#3.b.c Removing duplicates(Correcting inconsistent formats, Data Standardization(ISO 8601 format))
import pandas as pd

#Sample dataset with inconsistent data formats
df = pd.DataFrame({
    'Date': ['2025-01-05', '05/01/2025', 'Jan 5, 2025', '2025.01.05']
})
print("Original data:\n",df)

#Convert all dates to standard ISO format(YYYY-MM-DD)
df['Date'] = pd.to_datetime(df['Date'], errors='coerce').dt.strftime('%Y-%m-%d')
print(df)
```

```python
#3.b.d Removing duplicates(Correcting inconsistent formats, case normalization(lowercase/ uppercase))
import pandas as pd

#Sample dataset with inconsistent text cases
df = pd.DataFrame({
    'Name': ['Alice', 'BOB', 'CHarlIe', 'david'],
})
print("Original data:\n",df)

#convert to lowercase
df['Name_Lower'] = df['Name'].str.lower()

#Convert to Uppertcase
df['Name_Upper'] = df['Name'].str.upper()

print("\n",df)
```
